# Rough Filtering Users With Status

Status columns, unlike nutrition tag columns, indicate the specific health condition a user has. We hope to use this status information to focus on important and classic user groups.

```python
df_main = pd.read_csv('../processed_data/user_info_data.csv')

# Define the status columns
status_columns = ['Weight loss/Low calorie diet', 'Low fat/Low cholesterol diet',
                  'Low salt/Low sodium diet', 'Sugar free/Low sugar diet',
                  'Diabetic diet', 'Weight gain/Muscle building diet',
                  'Low carbohydrate diet', 'High protein diet', 'Renal/Kidney diet',
                  'obesity', 'hypertension', 'opioid_misuse', 'diabetes']
```

### Question 1: How many users does each stauts group has

```python
# Count the number of users for each status (value 1), ignoring NaN values
status_counts = df_main[status_columns].apply(lambda x: x.eq(1).sum(), axis=0)
# Display the result
print(status_counts)
```

### Question 2: For four major stauts, how many of them are on each diet?

```python
# Define the major status columns and the special diet columns
major_status_columns = ['obesity', 'hypertension', 'opioid_misuse', 'diabetes']
diet_columns = ['Weight loss/Low calorie diet', 'Low fat/Low cholesterol diet',
                'Low salt/Low sodium diet', 'Sugar free/Low sugar diet',
                'Diabetic diet', 'Weight gain/Muscle building diet',
                'Low carbohydrate diet', 'High protein diet', 'Renal/Kidney diet']

# Create a dictionary to store the counts
for status in major_status_columns:
    print(f"\nUsers with {status}:")
    # Filter users who have the major status (value is 1)
    filtered_users = df_main[df_main[status] == 1]
    
    # For each diet, count how many of these users are on that diet (diet value is 1)
    for diet in diet_columns:
        count = filtered_users[diet].value_counts().get(1, 0)  # Get count of users on the diet
        print(f"  {diet}: {count} users")
```

### Question 3: What does the overlaps between the four major status group look like?

```python
from itertools import combinations

# Define the major status columns
major_status_columns = ['obesity', 'hypertension', 'opioid_misuse', 'diabetes']

# Function to count users for a specific combination of statuses
def count_overlaps(status_list):
    # Filter rows where all columns in the status_list are equal to 1, ignoring NaN values
    return df_main.loc[:, status_list].eq(1).all(axis=1).sum()

# 6 scenarios for overlaps of 2 statuses
overlap_2_statuses = {}
for combo in combinations(major_status_columns, 2):
    overlap_count = count_overlaps(list(combo))  # Convert the tuple to a list
    overlap_2_statuses[f"Overlap between {combo[0]} and {combo[1]}"] = overlap_count

# 4 scenarios for overlaps of 3 statuses
overlap_3_statuses = {}
for combo in combinations(major_status_columns, 3):
    overlap_count = count_overlaps(list(combo))  # Convert the tuple to a list
    overlap_3_statuses[f"Overlap between {combo[0]}, {combo[1]}, and {combo[2]}"] = overlap_count

# 1 scenario for overlap of all 4 statuses
overlap_4_statuses = {}
overlap_count = count_overlaps(major_status_columns)
overlap_4_statuses["Overlap between all four statuses"] = overlap_count

# Display the results
print("Overlaps of 2 statuses:")
for key, value in overlap_2_statuses.items():
    print(f"{key}: {value} users")

print("\nOverlaps of 3 statuses:")
for key, value in overlap_3_statuses.items():
    print(f"{key}: {value} users")

print("\nOverlap of all 4 statuses:")
for key, value in overlap_4_statuses.items():
    print(f"{key}: {value} users")
```
