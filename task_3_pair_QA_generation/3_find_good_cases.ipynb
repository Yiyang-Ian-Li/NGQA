Load the data

```python
# load the pairs
import json

with open('processed_data/dedup_user_food_pair.json', 'r') as f:
    pair_list = json.load(f)
```

```python
# load the user info and the food info
import pandas as pd


user_tagging = pd.read_csv('../processed_data/user_tagging.csv')
food_tagging = pd.read_csv('../task_2_foods_filtering_qa_creation/processed_data/reduced_mixed_dishes_v4.csv')
```

```python
user_food_dict = {}

for food_user_dict in pair_list:

    keywords = ['easy', 'medium', 'hard']

    for keyword in keywords:
        for user_nutrition_dict in food_user_dict[keyword]:
            user_id = user_nutrition_dict['user_id']
            if user_id not in user_food_dict.keys():
                user_food_dict[user_id] = {'easy': [], 'medium': [], 'hard': []}
            user_food_dict[user_id][keyword].append(food_user_dict['food_id'])

# print(user_food_dict)
```
