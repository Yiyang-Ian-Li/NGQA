Load the data

```python
# load the pairs
import json

with open('processed_data/dedup_user_food_pair.json', 'r') as f:
    pair_list = json.load(f)
```

```python
# load the user info and the food info
import pandas as pd


user_tagging = pd.read_csv('../processed_data/user_tagging.csv').drop_duplicates()
food_tagging = pd.read_csv('../task_2_foods_filtering_qa_creation/processed_data/reduced_mixed_dishes_v4.csv')
```

```python
user_food_dict = {}

for food_user_dict in pair_list:

    keywords = ['easy', 'medium', 'hard']

    for keyword in keywords:
        for user_nutrition_dict in food_user_dict[keyword]:
            user_id = user_nutrition_dict['user_id']
            if user_id not in user_food_dict.keys():
                user_food_dict[user_id] = {'easy': [], 'medium': [], 'hard': []}
            user_food_dict[user_id][keyword].append(food_user_dict['food_id'])

# print(user_food_dict)
```

```python
# find users with both 'easy', 'medium', and 'hard' pairs and with more than 1 major status (only one user found)
for user_id in user_food_dict.keys():
    # if len(user_food_dict[user_id]['easy']) > 0 and len(user_food_dict[user_id]['medium']) > 0 and len(user_food_dict[user_id]['hard']) > 0:
    user_info = user_tagging[user_tagging['SEQN   '] == user_id]

    useful_coloum = ['obesity ', 'opioid_misuse ', 'hypertension ', 'diabetes ', 
                     'Weight loss/Low calorie diet ', 'Low fat/Low cholesterol diet ', 'Low salt/Low sodium diet ', 'Sugar free/Low sugar diet ', 'Diabetic diet ', 
                     'Weight gain/Muscle building diet ', 'Low carbohydrate diet ', 'High protein diet ', 'Renal/Kidney diet ']

    # print(user_info[useful_coloum].sum(axis=1))
    if user_info[useful_coloum].sum(axis=1).item() > 2 and user_info['Renal/Kidney diet '].item() == 1:
        print('User id:', user_info['SEQN   '].item())
        print(user_info[useful_coloum].iloc[0][user_info[useful_coloum].iloc[0] == 1].to_string(header=False))
        print('easy: ', [food_tagging[food_tagging['food_id'] == food_id]['food_desc'].item() for food_id in user_food_dict[user_id]['easy']])
        print('medium: ', [food_tagging[food_tagging['food_id'] == food_id]['food_desc'].item() for food_id in user_food_dict[user_id]['medium']])
        print('hard: ', [food_tagging[food_tagging['food_id'] == food_id]['food_desc'].item() for food_id in user_food_dict[user_id]['hard']])
        print('')
```
