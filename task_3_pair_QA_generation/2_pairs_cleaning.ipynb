## Generated pairs may contain duplicates so we need to do some cleaning

Load the pairs data

```python
# load the pairs
import json

with open('processed_data/food_based_pairs.json', 'r') as f:
    pair_list = json.load(f)
```

Original counts

```python
# original data
easy_sum = 0
medium_sum = 0
hard_sum = 0

for food_user_dict in pair_list:
    easy_sum += len(food_user_dict['easy'])
    medium_sum += len(food_user_dict['medium'])
    hard_sum += len(food_user_dict['hard'])

print(easy_sum, medium_sum, hard_sum)
```

Deduplicate

```python
# delete duplicates
easy_sum = 0
medium_sum = 0
hard_sum = 0
clean_pair_list = []

for food_user_dict in pair_list:
    food_user_dict['easy'] = [dict(t) for t in {json.dumps(d, sort_keys=True): d for d in food_user_dict['easy']}.values()]
    food_user_dict['medium'] = [dict(t) for t in {json.dumps(d, sort_keys=True): d for d in food_user_dict['medium']}.values()]
    food_user_dict['hard'] = [dict(t) for t in {json.dumps(d, sort_keys=True): d for d in food_user_dict['hard']}.values()]
    easy_sum += len(food_user_dict['easy'])
    medium_sum += len(food_user_dict['medium'])
    hard_sum += len(food_user_dict['hard'])

print(easy_sum, medium_sum, hard_sum)
```

Here we count the number of unique users and how many times they appear in 3 difficulty levels in our pairs

```python
user_dict = {}

for food_user_dict in pair_list:
    for user in food_user_dict['easy']:
        if user['user_id'] not in user_dict.keys():
            user_dict[user['user_id']] = {'easy': 0, 'medium':0, 'hard': 0}
        user_dict[user['user_id']]['easy'] += 1
    for user in food_user_dict['medium']:
        if user['user_id'] not in user_dict.keys():
            user_dict[user['user_id']] = {'easy': 0, 'medium':0, 'hard': 0}
        user_dict[user['user_id']]['medium'] += 1
    for user in food_user_dict['hard']:
        if user['user_id'] not in user_dict.keys():
            user_dict[user['user_id']] = {'easy': 0, 'medium':0, 'hard': 0}
        user_dict[user['user_id']]['hard'] += 1

# print(user_dict)
print(len(user_dict))

# for user in user_dict.keys():
#     if user_dict[user]['easy'] > 0 and user_dict[user]['medium'] > 0 and user_dict[user]['hard'] > 0:
#         print(user, user_dict[user])
```

Save the cleaned pairs

```python
import json

with open('processed_data/dedup_food_based_pairs.json', 'w') as f:
    json.dump(pair_list, f)
```
